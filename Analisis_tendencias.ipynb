```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Preparación de los datos
df_trends = pd.read_csv('interes_minecraft_movie.csv')

# Renombramos la columna para que sea más fácil llamarla
df_trends.rename(columns={'time [UTC]': 'Fecha'}, inplace=True)

# Convertimos la columna de texto a formato fecha real
df_trends['Fecha'] = pd.to_datetime(df_trends['Fecha'])

# 2. Configuración del estilo y tamaño
sns.set_theme(style="whitegrid")
plt.figure(figsize=(12, 6))

# 3. Creación del gráfico de líneas
sns.lineplot(data=df_trends, x='Fecha', y='Interes', color='#2ecc71', linewidth=2.5)

# 4. Personalización de textos
plt.title('Evolución del interés de búsqueda: "A Minecraft Movie"', fontsize=16, fontweight='bold', pad=20)
plt.xlabel('Fecha', fontsize=12, fontweight='bold')
plt.ylabel('Nivel de Interés (0-100)', fontsize=12, fontweight='bold')

# 5. Hitos del Caso (Líneas verticales)
# ⚠️ IMPORTANTE: Ajusta estas fechas según lo que investigues
fecha_estreno = pd.to_datetime('2025-04-04') # Fecha de estreno
fecha_meme = pd.to_datetime('2025-04-15')    # Fecha aproximada del meme

# Dibujamos la línea del estreno
plt.axvline(x=fecha_estreno, color='red', linestyle='--', alpha=0.8, linewidth=2)
plt.text(fecha_estreno, 80, ' Estreno en Cines', color='red', fontweight='bold', fontsize=11)

# Dibujamos la línea del meme viral
plt.axvline(x=fecha_meme, color='purple', linestyle='--', alpha=0.8, linewidth=2)
plt.text(fecha_meme, 60, ' Explosión Meme "Chicken Jockey"', color='purple', fontweight='bold', fontsize=11)

# 6. Mostrar el gráfico sin recortes
plt.tight_layout()
plt.show()
```
